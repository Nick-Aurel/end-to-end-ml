# Iris Dataset - Exploratory Data Analysis

This notebook explores the Iris dataset, which is used for training our classification model.

```python
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path('..').resolve()))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

from src.data import load_iris_data

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## Load Data

```python
# Load data using our custom function
X, y, target_names = load_iris_data()

print(f"Dataset shape: {X.shape}")
print(f"Number of classes: {len(target_names)}")
print(f"Class names: {target_names}")
```

## Create DataFrame for easier analysis

```python
# Get feature names
iris = load_iris()
feature_names = iris.feature_names

# Create DataFrame
df = pd.DataFrame(X, columns=feature_names)
df['target'] = y
df['species'] = df['target'].map({i: name for i, name in enumerate(target_names)})

df.head()
```

## Basic Statistics

```python
df.describe()
```

## Class Distribution

```python
df['species'].value_counts().plot(kind='bar')
plt.title('Class Distribution')
plt.xlabel('Species')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

## Pairwise Feature Relationships

```python
sns.pairplot(df, hue='species', vars=feature_names, height=2.5)
plt.suptitle('Pairwise Feature Relationships', y=1.02)
plt.show()
```
